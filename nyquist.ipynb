```python
def nyquist_table(fs, f_sig, f_upper=None, max_harmonic=5):
    """
    fs        : sampling rate (Hz/MHz/GHz — units are implicit)
    f_sig     : fundamental signal frequency
    f_upper   : upper frequency bound for display (default: 5 * fs)
    max_harmonic : highest harmonic order to consider
    """
    if f_upper is None:
        f_upper = 5 * fs
 
    rows = []
 
    for h in range(1, max_harmonic + 1):
        f_harm = h * f_sig
        h_label = f"{h}f" if h > 1 else "f"
 
        # Fundamental / harmonic itself
        if f_harm <= f_upper:
            rows.append((f_harm, f"${h_label}$", f"${h}×f = {f_harm}$"))
 
        # Nyquist images:  n·fs ± h·f_sig  for n = 1, 2, 3, ...
        n = 1
        while True:
            nfs = n * fs
            if nfs - f_harm > f_upper and nfs + f_harm > f_upper:
                break
 
            f_plus  = nfs + f_harm
            f_minus = abs(nfs - f_harm)
 
            if f_plus <= f_upper:
                rows.append((f_plus,
                             f"${n}f_s + {h_label}$",
                             f"${n}×{fs} + {h}×{f_sig} = {f_plus}$"))
            if f_minus <= f_upper:
                sign = "−" if nfs >= f_harm else "+"
                rows.append((f_minus,
                             f"${n}f_s {sign} {h_label}$",
                             f"$|{n}×{fs} − {h}×{f_sig}| = {f_minus}$"))
            n += 1
 
    # Sort by frequency and deduplicate
    rows.sort(key=lambda r: r[0])
    seen = set()
    unique = []
    for freq, label, expr in rows:
        if freq not in seen:
            seen.add(freq)
            unique.append((freq, label, expr))
 
    # Pretty print
    col1, col2, col3 = "Freq", "Label", "Expression"
    w1 = max(len(str(r[0])) for r in unique) + 2
    w2 = max(len(r[1]) for r in unique) + 2
    w3 = max(len(r[2]) for r in unique) + 2
 
    header = f"{'Freq':<{w1}}  {'Label':<{w2}}  {'Expression':<{w3}}"
    print(f"\n  f_sig = {f_sig},  f_s = {fs},  upper bound = {f_upper}\n")
    print(f"  {header}")
    print(f"  {'─'*len(header)}")
    for freq, label, expr in unique:
        print(f"  {str(freq):<{w1}}  {label:<{w2}}  {expr:<{w3}}")
    print()
```

```python
nyquist_table(fs=9.584, f_sig=4, f_upper=16, max_harmonic=5)
```
